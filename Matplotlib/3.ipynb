```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([1,8])
y = np.array([3,10])
plt.plot(x, y,'x')
plt.show()
```

```python
x = np.array([2,2,2,2,10,10,10,10,4,7,4,7])
y = np.array([3,5,7,9,3,5,7,9,9,9,3,3])
plt.plot(x, y,'x')
plt.show()
```

```python
x = np.array([1,1,1,1,2,3,4,4,4,4,3,2,1])
y = np.array([3,4,5,6,6,6,6,5,4,3,3,3,3])
plt.plot(x, y, marker = 'o')
plt.show()
```

```python
x = np.array([1,1,1,1,2,3,4,4,4,4,3,2,1])
y = np.array([3,4,5,6,6,6,6,5,4,3,3,3,3])
plt.plot(x, y, marker = 'o',color='red')
plt.show()
```

```python
x = np.array([1,1,1,1,2,3,4,4,4,4,3,2,1])
y = np.array([3,4,5,6,6,6,6,5,4,3,3,3,3])
plt.plot(x, y,'o:k') # k is used for black color
plt.show()
```

```python
x1 = np.array([0,1,2,3])
y1 = np.array([3,8,1,10])
x2 = np.array([0,1,2,3])
y2 = np.array([6,2,7,11])
plt.plot(x1, y1, label = 'line 1')
plt.plot(x2, y2, label = 'line 2')
plt.legend()
```
