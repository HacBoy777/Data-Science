```python
import matplotlib.pyplot as plt
import pandas as pd 
import numpy as np
```

Create a Bar Chart for MAANG(Meta,Apple,Amazon,Netflix,Google) company with its wealth 

```python
company = np.array(['Meta', 'Apple', 'Amazon', 'Netflix', 'Google'])
wealth = np.array([15000, 25000, 17000, 30000, 22000])
plt.bar(company, wealth, color ='green')
plt.show()
```

1. X Column with company name and label as maang
2. Y column with wealth and label as wealth (Billon)
3. Give title as MAANG Company Stocks
4. Give legend as wealth details

```python
plt.title('MAANG Company Stocks')
plt.xlabel('MAANG')
plt.ylabel('Wealth (Billion)')
plt.bar(company, wealth, color ='green',label='Wealth Details')
plt.legend()
plt.show()
```
