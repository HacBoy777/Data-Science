```python
##############Task-I-understand following code#######################
## hotel is identifying which fruit how much consumed
import matplotlib.pyplot as plt
import numpy as np
fruit = np.array([35, 25, 25, 15])
# f_label = np.array(["Grapes","Orange","Kiwi","Apple"])
plt.pie(fruit)
plt.show() 
```

```python
# Task-2 ## give label to fruit in pie chart #
f_label = ["Grapes","Orange","Kiwi","Apple"]
plt.title("Fruits Consumed in Hotel")
plt.pie(fruit,labels=f_label)
plt.show() 
```

```python
# Task-3 ## give color to each fruit in pie chart #
fruitlbl = ["Cherry", "Banana", "Watermellon", "Orange"]
plt.pie(fruit,labels=fruitlbl,colors=['r','yellow','g','orange'])
plt.show()
```

```python
# Task-4 give legend in pie chart as "Fruit consumed and explode cherry"
fruitmostconsumed = [0.2, 0, 0, 0]
plt.pie(fruit,labels=fruitlbl,colors=['r','yellow','g','orange'],explode = fruitmostconsumed)
plt.legend(title="Fruits Consumed",loc='upper left')
plt.show()
```
