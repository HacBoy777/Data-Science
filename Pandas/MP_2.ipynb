```python
files = os.listdir("dataset/mp")
for file in files:
    if file.endswith(".txt"):
        path = os.path.join("dataset/mp", file)
        os.remove(path)
        print(f"Deleted: {file}")
```

```python
if os.path.exists("Dataset/MP/all_sales.csv"):
  os.remove("Dataset/MP/all_sales.csv")
else:
  print("The file does not exist")
```

```python
folder = "dataset/mp"
csv_files = [f for f in os.listdir(folder) if f.endswith(".csv")]
dataframes = []
for file in csv_files:
    path = os.path.join(folder, file)
    df = pd.read_csv(path)
    df["source_file"] = file  
    dataframes.append(df)
all_sales = pd.concat(dataframes, ignore_index=True)
print("all_sales DataFrame:")
print(all_sales)
```
