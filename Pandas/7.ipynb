```python
import pandas as pd 
maharstra_weather = pd.DataFrame({
    "city" : ["Mumbai","Pune","Nagpur"],
    "temprature" : [32,30,40],
    "humidity" : [60,70,80]
})
maharstra_weather
```

```python
gujrat_weather=pd.DataFrame({
"city":["Baroda","Godhra","Bharuch"],
"temprature":[32,40,30],
"humidity":[80,60,70]
})
gujrat_weather
```

```python
df = pd.concat([gujrat_weather,maharstra_weather])
df
```

```python
df = pd.concat([gujrat_weather,maharstra_weather],ignore_index=True)
df
```

```python
df = pd.concat([gujrat_weather,maharstra_weather],keys=["Gujarat","Maharastra"])
df
```
