```python
import pandas as pd
df = pd.read_csv("Dataset\\weather_data.csv")
df
```

```python
# Fill NaN by ZERO
new_df = df.fillna(0)
new_df 
```

```python
new_df = df.fillna({
    'temperature' : 0,
    'windspeed' :'No wind',
    'event' : 'No event' 
})
new_df
```

```python
# Forward fill
new_df = df.ffill()
new_df 
```

```python
# Backward fill
new_df = df.bfill()
new_df 
```

```python
# Change data row wise (backward fill)
new_df = df.bfill(axis = 'columns')
new_df
```

```python
# Change data row wise (forward fill)
new_df = df.bfill(axis = 'rows')
new_df
```

```python
# Drop NaN values
new_df = df.dropna()
new_df
```
